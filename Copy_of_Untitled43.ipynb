```python
import pandas as pd
import numpy as np
from scipy.stats import linregress
import os

os.makedirs("outputs", exist_ok=True)

# ==================================================
# LOAD DATA
# ==================================================

nav = pd.read_csv("clean_nav.csv")
benchmark = pd.read_csv("10_benchmark_indices.csv")

nav['date'] = pd.to_datetime(nav['date'])
benchmark['date'] = pd.to_datetime(benchmark['date'])

# ==================================================
# DAILY RETURNS
# ==================================================

nav = nav.sort_values(
    ['amfi_code','date']
)

nav['daily_return'] = (
    nav.groupby('amfi_code')['nav']
    .pct_change()
)

# ==================================================
# CAGR CALCULATION
# ==================================================

cagr_list = []

for fund in nav['amfi_code'].unique():

    temp = nav[
        nav['amfi_code'] == fund
    ]

    start_nav = temp['nav'].iloc[0]
    end_nav = temp['nav'].iloc[-1]

    years = (
        temp['date'].max()
        -
        temp['date'].min()
    ).days / 365

    cagr = (
        (end_nav/start_nav)
        ** (1/years)
        - 1
    ) * 100

    cagr_list.append([
        fund,
        round(cagr,2)
    ])

cagr_df = pd.DataFrame(
    cagr_list,
    columns=[
        'amfi_code',
        'cagr_pct'
    ]
)

# ==================================================
# SHARPE RATIO
# ==================================================

risk_free_rate = 0.065

sharpe_results = []

for fund in nav['amfi_code'].unique():

    returns = nav[
        nav['amfi_code']==fund
    ]['daily_return'].dropna()

    annual_return = (
        returns.mean() * 252
    )

    annual_volatility = (
        returns.std()
        * np.sqrt(252)
    )

    sharpe = (
        annual_return
        -
        risk_free_rate
    ) / annual_volatility

    sharpe_results.append([
        fund,
        sharpe
    ])

sharpe_df = pd.DataFrame(
    sharpe_results,
    columns=[
        'amfi_code',
        'sharpe_ratio'
    ]
)

# ==================================================
# SORTINO RATIO
# ==================================================

sortino_results = []

for fund in nav['amfi_code'].unique():

    returns = nav[
        nav['amfi_code']==fund
    ]['daily_return'].dropna()

    downside = returns[
        returns < 0
    ]

    downside_std = (
        downside.std()
        * np.sqrt(252)
    )

    annual_return = (
        returns.mean()
        * 252
    )

    sortino = (
        annual_return
        -
        risk_free_rate
    ) / downside_std

    sortino_results.append([
        fund,
        sortino
    ])

sortino_df = pd.DataFrame(
    sortino_results,
    columns=[
        'amfi_code',
        'sortino_ratio'
    ]
)

# ==================================================
# ALPHA & BETA
# ==================================================

benchmark = benchmark.rename(
    columns={
        'close_value':'benchmark_close' # Corrected column name
    }
)

benchmark['benchmark_return'] = (
    benchmark['benchmark_close']
    .pct_change()
)

alpha_beta = []

for fund in nav['amfi_code'].unique():

    temp = nav[
        nav['amfi_code']==fund
    ][['date','daily_return']]

    merged = pd.merge(
        temp,
        benchmark[
            ['date',
             'benchmark_return']
        ],
        on='date',
        how='inner'
    )

    merged = merged.dropna()

    if len(merged) > 30:

        slope, intercept, _, _, _ = (
            linregress(
                merged['benchmark_return'],
                merged['daily_return']
            )
        )

        beta = slope

        alpha = (
            intercept
            * 252
            * 100
        )

        alpha_beta.append([
            fund,
            alpha,
            beta
        ])

alpha_beta_df = pd.DataFrame(
    alpha_beta,
    columns=[
        'amfi_code',
        'alpha',
        'beta'
    ]
)

# ==================================================
# MAX DRAWDOWN
# ==================================================

drawdown_results = []

for fund in nav['amfi_code'].unique():

    temp = nav[
        nav['amfi_code']==fund
    ]

    running_max = (
        temp['nav']
        .cummax()
    )

    drawdown = (
        temp['nav']
        /
        running_max
        - 1
    )

    max_dd = (
        drawdown.min()
        * 100
    )

    drawdown_results.append([
        fund,
        max_dd
    ])

drawdown_df = pd.DataFrame(
    drawdown_results,
    columns=[
        'amfi_code',
        'max_drawdown_pct'
    ]
)

# ==================================================
# MERGE ALL METRICS
# ==================================================

performance = cagr_df.merge(
    sharpe_df,
    on='amfi_code'
)

performance = performance.merge(
    sortino_df,
    on='amfi_code'
)

performance = performance.merge(
    alpha_beta_df,
    on='amfi_code'
)

performance = performance.merge(
    drawdown_df,
    on='amfi_code'
)

# ==================================================
# FUND SCORECARD
# ==================================================

performance['cagr_rank'] = (
    performance['cagr_pct']
    .rank(
        ascending=False
    )
)

performance['sharpe_rank'] = (
    performance['sharpe_ratio']
    .rank(
        ascending=False
    )
)

performance['alpha_rank'] = (
    performance['alpha']
    .rank(
        ascending=False
    )
)

performance['drawdown_rank'] = (
    performance['max_drawdown_pct']
    .rank(
        ascending=False
    )
)

performance['fund_score'] = (
    performance['cagr_rank']*0.30
    +
    performance['sharpe_rank']*0.25
    +
    performance['alpha_rank']*0.20
    +
    performance['drawdown_rank']*0.25
)

performance = performance.sort_values(
    'fund_score'
)

# ==================================================
# SAVE OUTPUTS
# ==================================================

performance.to_csv(
    "outputs/fund_scorecard.csv",
    index=False
)

cagr_df.to_csv(
    "outputs/cagr_report.csv",
    index=False
)

alpha_beta_df.to_csv(
    "outputs/alpha_beta.csv",
    index=False
)

drawdown_df.to_csv(
    "outputs/max_drawdown.csv",
    index=False
)

print("\nDAY 4 COMPLETED")
performance.head()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

os.makedirs("outputs", exist_ok=True)

# ============================================
# LOAD DATA
# ============================================

nav = pd.read_csv("clean_nav.csv")

transactions = pd.read_csv(
    "clean_transactions.csv"
)

portfolio = pd.read_csv(
    "09_portfolio_holdings.csv"
)

performance = pd.read_csv(
    "outputs/fund_scorecard.csv"
)

nav['date'] = pd.to_datetime(
    nav['date']
)

transactions['transaction_date'] = pd.to_datetime(
    transactions['transaction_date']
)

# ============================================
# DAILY RETURNS
# ============================================

nav = nav.sort_values(
    ['amfi_code','date']
)

nav['daily_return'] = (
    nav.groupby('amfi_code')['nav']
    .pct_change()
)

# ============================================
# TASK 1 - VaR & CVaR
# ============================================

var_results = []

for fund in nav['amfi_code'].unique():

    returns = nav[
        nav['amfi_code']==fund
    ]['daily_return'].dropna()

    var95 = np.percentile(
        returns,
        5
    )

    cvar95 = returns[
        returns <= var95
    ].mean()

    var_results.append([
        fund,
        var95,
        cvar95
    ])

var_df = pd.DataFrame(
    var_results,
    columns=[
        'amfi_code',
        'VaR_95',
        'CVaR_95'
    ]
)

var_df.to_csv(
    "outputs/var_cvar_report.csv",
    index=False
)

# ============================================
# TASK 2 - Rolling Sharpe Ratio
# ============================================

selected_fund = nav[
    'amfi_code'
].iloc[0]

fund_data = nav[
    nav['amfi_code']
    ==
    selected_fund
].copy()

rolling_mean = (
    fund_data['daily_return']
    .rolling(90)
    .mean()
)

rolling_std = (
    fund_data['daily_return']
    .rolling(90)
    .std()
)

rolling_sharpe = (
    rolling_mean
    /
    rolling_std
) * np.sqrt(252)

plt.figure(figsize=(12,6))

plt.plot(
    fund_data['date'],
    rolling_sharpe
)

plt.title(
    "90-Day Rolling Sharpe Ratio"
)

plt.savefig(
    "outputs/rolling_sharpe_chart.png"
)

plt.close()

# ============================================
# TASK 3 - Cohort Analysis
# ============================================

transactions['cohort_year'] = (
    transactions.groupby(
        'investor_id'
    )['transaction_date']
    .transform('min')
    .dt.year
)

cohort_analysis = (
    transactions
    .groupby('cohort_year')
    .agg({
        'amount_inr':'mean',
        'investor_id':'nunique'
    })
    .reset_index()
)

cohort_analysis.to_csv(
    "outputs/cohort_analysis.csv",
    index=False
)

# ============================================
# TASK 4 - SIP Continuity Analysis
# ============================================

sip = transactions[
    transactions[
        'transaction_type'
    ]
    ==
    'SIP'
].copy()

sip = sip.sort_values(
    ['investor_id',
     'transaction_date']
)

sip['gap_days'] = (
    sip.groupby(
        'investor_id'
    )['transaction_date']
    .diff()
    .dt.days
)

at_risk = sip[
    sip['gap_days'] > 35
]

at_risk.to_csv(
    "outputs/sip_continuity.csv",
    index=False
)

# ============================================
# TASK 5 - Recommendation Engine
# ============================================

def recommend_funds(risk_level):

    if risk_level == "Low":

        result = performance.nsmallest(
            3,
            'max_drawdown_pct'
        )

    elif risk_level == "Moderate":

        result = performance.nlargest(
            3,
            'sharpe_ratio'
        )

    else:

        result = performance.nlargest(
            3,
            'cagr_pct'
        )

    return result[
        [
            'amfi_code',
            'cagr_pct',
            'sharpe_ratio'
        ]
    ]

print(
    "\nModerate Risk Recommendation"
)

print(
    recommend_funds("Moderate")
)

# ============================================
# TASK 6 - Sector Concentration Risk
# ============================================

sector_hhi = (
    portfolio
    .groupby(
        ['amfi_code','sector']
    )['weight_pct']
    .sum()
    .reset_index()
)

hhi_results = []

for fund in sector_hhi[
    'amfi_code'
].unique():

    temp = sector_hhi[
        sector_hhi[
            'amfi_code'
        ]
        ==
        fund
    ]

    weights = (
        temp['weight_pct']
        / 100
    )

    hhi = np.sum(
        weights**2
    )

    hhi_results.append([
        fund,
        hhi
    ])

hhi_df = pd.DataFrame(
    hhi_results,
    columns=[
        'amfi_code',
        'HHI'
    ]
)

hhi_df.to_csv(
    "outputs/sector_hhi.csv",
    index=False
)

# ============================================
# TASK 7 - Advanced Insights
# ============================================

with open(
    "outputs/advanced_insights.txt",
    "w"
) as file:

    file.write(
        "ADVANCED ANALYTICS INSIGHTS\n\n"
    )

    file.write(
        f"Total Funds: {len(var_df)}\n"
    )

    file.write(
        f"At-Risk SIP Investors: {len(at_risk)}\n"
    )

    file.write(
        f"Total Cohorts: {cohort_analysis.shape[0]}\n"
    )

    file.write(
        f"Highest HHI: {hhi_df['HHI'].max()}\n"
    )

print("\nDAY 6 COMPLETED")
print("Outputs saved successfully")
```
